```python
# 제네레이터
# 함수 자체가 순서대로 하나씩 내어보내도록

def numbers():
    yield 1
    yield 2
    yield 3

numbers()
for n in numbers():
    print(n)
```

```python
# yield : 잠시 멈췄다가 양보하고 이어가는 형태
def demo():
    print("a")
    yield 1
    print("b")
    yield 2
    print("c")
    yield 3
    print("d")
 
```
